# Training on Kaggle (GPU)

1. Zip the repo (`msdemosaic/`, `pyproject.toml`) and upload it as a Kaggle dataset, e.g. `msdemosaic-code`.
2. Upload the data as a second dataset with the layout `CAVE/<scene>_ms/...` (and `ARAD_1K/Train_Spec/*.mat` for `--data-mode arad`).
3. Enable the GPU accelerator, set the three paths below and run all cells.
4. Download `/kaggle/working/checkpoints/*` into the local `checkpoints/` folder.

```python
CODE = '/kaggle/input/msdemosaic-code'
DATA = '/kaggle/input/msdemosaic-data'
MODE = 'cave-split'  # or 'arad'
ITERS = 40000
```

```python
# Kaggle inputs are read-only: copy the code, and link the data so the cache can be written.
!cp -r {CODE} /kaggle/working/repo && pip install -q -e /kaggle/working/repo
!mkdir -p /kaggle/working/data && ln -sfn {DATA}/CAVE /kaggle/working/data/CAVE
!test -d {DATA}/ARAD_1K && ln -sfn {DATA}/ARAD_1K /kaggle/working/data/ARAD_1K; true
```

```python
%cd /kaggle/working/repo
for pattern, name in [('imec', 'fixed_imec'), ('osp', 'fixed_osp'), ('random', 'agnostic')]:
    !python -m msdemosaic.train --pattern {pattern} --data-root /kaggle/working/data --data-mode {MODE} --iters {ITERS} --out /kaggle/working/checkpoints/{name}.pt
```
